# Reaction-Diffusion Systems

$$\frac{\partial u}{\partial t} = f(u, v) + D_u \nabla^2 u$$
$$\frac{\partial v}{\partial t} = g(u, v) + D_v \nabla^2 v$$
$f(u, v)$ and $g(u, v)$: Non-linear reaction terms defining the oscillator kinetics.
$D_u$ and $D_v$: Diffusion coefficients governing how fast each species spreads.
$\nabla^2$: The Laplacian operator, which computes local spatial curvature in grids

$$\nabla^2 u = \frac{\partial^2 u}{\partial x^2} + \frac{\partial^2 u}{\partial y^2}$$
Sum of partial derivatives in 2D shortcut: Top + Bottom + Left + Right - 4 * Center

The Brusselator Example: 
Reaction Scheme

$$A \xrightarrow{} X$$

$$2X + Y \xrightarrow{} 3X$$

$$B + X \xrightarrow{} Y + D$$

$$X \xrightarrow{} E$$

A theoretical four-step reaction that resulted in famous pair of equations:

$f(u, v) = A - (B + 1)u + u^2v$

$g(u, v) = Bu - u^2v$

Here, $A$ and $B$ are constant inputs of chemicals, while $u^2v$ represents the non-linear loop where $u$ feeds into $v$. 
These equations don't describe a specific real-world chemical process, but capture chemical oscillators behaviour.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Parameters
N = 100          # Grid size (N x N)
dx = 1.0         # Space step
dt = 0.01        # Time step
D_u = 1.0        # Diffusion coefficient for u
D_v = 8.0        # Diffusion coefficient for v (v must diffuse faster than u!)
A = 1.0
B = 3.0          # Brusselator kinetic parameters

# Initialize Grids with equilibrium (u - concentration of the first chemical species, v - concentration of the first chemical species)
u = A + np.random.rand(N, N) * 0.1
v = B / A + np.random.rand(N, N) * 0.1

def get_laplacian(temp):
    return (np.roll(temp, 1, axis=0) + np.roll(temp, -1, axis=0) + 
            np.roll(temp, 1, axis=1) + np.roll(temp, -1, axis=1) - 4 * temp) / (dx ** 2)

fig, ax = plt.subplots(figsize=(6, 6))
im = ax.imshow(u, cmap='inferno', animated=True)
ax.axis('off')
plt.title('Brusselator Reaction-Diffusion Animation')

def update(frame):
    global u, v
    
    for i in range(5):
        lap_u = get_laplacian(u)
        lap_v = get_laplacian(v)
        
        f = A - (B + 1)*u + (u**2)*v
        g = B*u - (u**2)*v
        
        u += (f + lap_u*D_u)*dt
        v += (g + lap_v*D_v)*dt

    im.set_array(u)
    return [im]

anim = FuncAnimation(fig, update, frames=150, interval=30, blit=True)
anim.save('brusselator.gif', writer='pillow', fps=30)
plt.close()

HTML(anim.to_jshtml())
```
